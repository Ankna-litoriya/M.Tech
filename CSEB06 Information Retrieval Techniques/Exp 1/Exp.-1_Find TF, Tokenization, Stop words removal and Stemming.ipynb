```python
import nltk
import pandas as pd
from docx import Document
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from collections import Counter
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
file_path = "big_data_technical_content_3_pages.docx"

doc = Document(file_path)

text = ""

for para in doc.paragraphs:
    text += para.text + " "

print("File Loaded Successfully\n")
print(text[:500])
```

```python
tokens = word_tokenize(text)

print("Total Tokens:", len(tokens))

print("\nFirst 50 Tokens:\n")
print(tokens[:50])
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word.lower() for word in tokens if word.lower() not in stop_words and word.isalpha()]

print("Total Words After Stopword Removal:", len(filtered_tokens))

print("\nFirst 50 Words After Stopword Removal:\n")
print(filtered_tokens[:50])
```

```python
stemmer = PorterStemmer()

sample_words = filtered_tokens[:10]

stemmed_words = []

for word in sample_words:
    stemmed_words.append(stemmer.stem(word))

print("Original Words:\n", sample_words)

print("\nStemmed Words:\n", stemmed_words)
```

```python
technical_terms = ["data","hadoop","spark","kafka","processing"]

term_frequency = {}

for term in technical_terms:
    term_frequency[term] = filtered_tokens.count(term)

print("Technical Term Frequency:\n")
print(term_frequency)
```

```python
ranked_terms = sorted(term_frequency.items(), key=lambda x: x[1], reverse=True)

print("Ranking of Technical Terms:\n")

for rank,(term,freq) in enumerate(ranked_terms,start=1):
    print(rank,".",term,"=",freq)
```

```python
df = pd.DataFrame(ranked_terms, columns=["Technical Term","Frequency"])

df["Rank"] = df["Frequency"].rank(ascending=False, method="dense").astype(int)

print("\nTechnical Term Ranking Table\n")

print(df)
```
